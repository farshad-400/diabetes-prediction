```python
X = df.drop('diabetes', axis=1)
y = df['diabetes']
```

```python
y
```

```python
X
```

```python
over=SMOTE(random_state=1)
X,y=over.fit_resample(X,y)
```

```python
y.value_counts()
```

```python
n=MinMaxScaler()
X=n.fit_transform(X)
```

```python
X
```

```python
  X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3,random_state=1)
```

```python
X_train.shape, X_test.shape
```

```python
y_train.shape,y_test.shape
```

```python
k=KNeighborsClassifier()
r=RandomForestClassifier()
d=DecisionTreeClassifier()
a=AdaBoostClassifier()
g=GaussianNB()
b=BernoulliNB()
m=MultinomialNB()
f=SVC()
e=LogisticRegression()
model=[k,r,d,a,g,b,m,f,e]

for i in model:
    print('',i)
    print('********************************************************************************************')
    i.fit(X_train,y_train)
    y_pred=i.predict(X_test)
    print('accuracy=', accuracy_score(y_test,y_pred)*100)
    print(classification_report(y_test,y_pred))
    print(ConfusionMatrixDisplay.from_predictions(y_test,y_pred))
    print()
    print()
```

```python
from sklearn.model_selection import cross_val_score
print("Cross Validation Scores for all models on training data:")
for i in model:
    print("        ", i.__class__.__name__)
    print("*****************************************************************")
    # Using cv-3 and n_jobs-1 for faster execution.
    scores = cross_val_score(i, X, y, cv=3, scoring= 'accuracy', n_jobs=-1)
    print (f"CV Accuracy Scores: {scores}")
    print(F"Mean Accuracy: {np.mean(scores)*100:.2f}%\n")
```
